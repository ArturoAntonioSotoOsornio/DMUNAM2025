```python
import pandas as pd
```

Leer archivos desde mi Drive y cargar los archivos

```python
ruta = "/content/drive/MyDrive/Practica1"

cb = pd.read_csv("/content/drive/MyDrive/Practica1/afluencia_desglosada_cb_03_2025.csv")
tl = pd.read_csv("/content/drive/MyDrive/Practica1/afluencia_desglosada_tl_03_2025.csv")
trolebus = pd.read_csv("/content/drive/MyDrive/Practica1/afluencia_desglosada_trolebus_03_2025.csv")
```

Tipo de transporte

```python
cb["tipo"] = "Cablebús"
tl["tipo"] = "Tren Ligero"
trolebus["tipo"] = "Trolebús"
```

Concatenar en un solo dataframe

```python
df_total = pd.concat([cb, tl, trolebus], ignore_index=True)
df_total.head()
```

```python
print(df_total.columns.tolist())
```

Merge usando tipo_pago

```python
df_pagos = pd.DataFrame({
    'tipo_pago': ['Prepago', 'Gratuita'],
    'descripcion': ['Pago con tarjeta', 'Acceso libre']
})

df_merged = pd.merge(df_total, df_pagos, on='tipo_pago', how='left')
df_merged.head()
```

Afluencia total por tipo de pago

```python
df_merged.groupby('tipo_pago')['afluencia'].sum().reset_index()
```

Afluencia total por descripción del pago

```python
df_merged.groupby('descripcion')['afluencia'].sum().reset_index()
```

Afluencia por tipo de transporte y tipo de pago

```python
df_merged.groupby(['tipo', 'tipo_pago'])['afluencia'].sum().reset_index()
```

Promedio de afluencia por tipo de pago

```python
df_merged.groupby('tipo_pago')['afluencia'].mean().reset_index()
```
