```python
import pandas as pd
import glob
import os
```

```python

ruta = "C:/Users/pablo/OneDrive/Documentos/FES/Ecobici datos/2024"

# Obtener todos los archivos CSV en la carpeta
csv_files = [os.path.join(ruta, f) for f in os.listdir(ruta) if f.lower().endswith(".csv")]
```

```python

# Leer y combinar todos
dfs = [pd.read_csv(f, low_memory=False) for f in csv_files]

# Unirlos
df_combined = pd.concat(dfs, ignore_index=True)
```

```python

# Guardar resultado
df_combined.to_csv("ecobici2024.csv", index=False)

print("¡Archivos combinados exitosamente!")
```
